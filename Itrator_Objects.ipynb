### What are Iterators?

* An object helps in memory management
* Iterators are used to access the items in a collection object without exposing the underlying concept.

### How does they work? Difference in Generators, Iterator and Iterables?

* Like a list, loads the whole collection in RAM instantly at a time where as iterators are more efficient in memory management, iterators load items one-by-one when requested. 
* Iterators uses __ next __ to iterator over the collection, allows to access items one-by-one.

* Generators use yield keyword to generate item at time, internally they use __ next __ method.
* Iterables are the collection but they cannot be a stream. Every iterable have __ next __ and __ iter __ method that allows them to be a stream when they are looped.
* Iterators convert the iterables to a stream, so, items can be accessed one at a time.

### Why we don't use __ next __() in loop?

When we access the item using the next() function.
 - First when next() used the first item returned.
 - Second when next() used, __ iter __() called internally to access the next item.
 - Same process is looped whenever we call the next().

In for loop we did not call the next() function. In loops __ iter __() and __ next __() are called internally. 

### How does it works?

Custom iterator

```python
class MyCustomIterator:
    def __init__(self,start):
    # No. of times you iterate
        self.current = start

    def __iter__(self):
    # iter() return itself for the next iteration
        return self

    def __next__(self):
    # Access the next item in the sequence
        if self.current<=0:
            raise StopIteration
        self.current-=1
        return self.current

iterator=MyCustomIterator(5)
print(iterator)
print(next(iterator))
print(iterator.__next__())
for item in iterator:
    print(item)
```

### What happen if return something else?

```python
class MyCustomIterator:
    def __init__(self,start):
    # No. of times you iterate
        self.current = start

    def __iter__(self):
    # iter() return the number not itself
        return self.current

    def __next__(self):
    # Access the next item in the sequence
        if self.current<=0:
            raise StopIteration
        self.current-=1
        return self.current

iterator=MyCustomIterator(5)
print(iterator)
print(next(iterator))
print(iterator.__next__())
for item in iterator:
    print(item)
```

```python
class MyCustomIterator:
    def __init__(self,start):
    # No. of times you iterate
        self.current = start

    def __iter__(self):
    # iter() return the range
        return list(range(self.current))

    def __next__(self):
    # Access the next item in the sequence
        if self.current<=0:
            raise StopIteration
        self.current-=1
        return self.current

iterator=MyCustomIterator(5)
for item in iterator:
    print(item)
```

Self observation - __ iter __(), when return self (itself object) means returning the same collection object to iterate over the next item. It only allows to return the object refernce.

**Create a custom iterator for collection object**

```python
class Iterator:
    def __init__(self,collection):
        self.iterable=collection
        self.count=0

    def __iter__(self):
        return self

    def __next__(self):
        try:
            item = self.iterable[self.count]
            self.count+=1
        except Exception as e:
            raise StopIteration(e)
        return item
x=[87,38,3,93,7,51,2]
list_object = Iterator(x)
for item in list_object:
    print(item)
```
