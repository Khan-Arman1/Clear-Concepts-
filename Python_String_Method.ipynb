```python
# 11. find() - specific string/single char ko find karta hai. 
# Index() ki tarha hai but index() value na milne par error raise karta hai.
# find() value na milne par '-1 return' karta hai
# value milne par us specific character sequence ki 'first character ka index return karega'
# 3 parameter leta hai
txt = "Hello, I am Arman Khan."
x = txt.find("khan")
print(x)
x = txt.find("han",
            15, # start index
             len(txt) # end index
            )
print(x)
```

```python
# 16. isalnum() - Check karta hai sare characters alpha numeric hai ya nahi, alphate[A-z] or numeric [0-9]
# Agar hota hai toh True return karta hai.
# No parameter
# Characters that are not aplhanumeric: (space)!#%&?
txt = "I am a Python Programmer and Data Science Learner."
x = txt.isalnum()
print(x)
txt = "I am a Python Programmer for 1.5 years and Data Science Learner."
x = txt.isalnum()
print(x)
txt = "arman12"
x = txt.isalnum()
print(x)
txt = "arman12@gmail.com"
x = txt.isalnum()
print(x)

# Insights - Koi bhi special character hota hai toh False return karta hai (bola gaya tha) only alphabatic characters and numbers hi check karega
```

```python
# 17. isalpha() - Return True agar sare sara characters aphabates hai toh (a-z,A-Z)
# special characters aphabats nahi hote.
# No parameters
txt = "I am a Python Programmer."
x = txt.isalpha()
print(x) # expected output False because string have whitespaces
txt = "Arman_Khan."
x = txt.isalpha()
print(x) # expected output False because string have _ & .
txt = "ArmanKhan"
x = txt.isalpha()
print(x) # expected output True
txt = "ArmanKhan1"
x = txt.isalpha()
print(x) # expected output False
```

```python
# 18. isascii() - REturn True agar sare characters ascii characters hai toh.
# Keyboard ke sare characters ascii characters me ate hai.
# No parameter
txt  = "Hello, I am a Data Analyst."
x = txt.isascii()
print(x) # expected output True all are ascii
txt  = "armankhan52210@gmail.com"
x = txt.isascii()
print(x) # expected output True all are ascii
txt  = "Some What character that is not ascii €."
x = txt.isascii()
print(x) # expected output False all are not ascii
```

```python
# 19. isdecimal() - Return True agar sare number hai toh (0-9).
# Unicode characters par bhi work karta hai. Like G,A ko aap unicode me likhte ho.
# (unicode - encoding hi hoti hai buss jo characters ko samjhne ke liye use hoti hai)
# no parameter
txt = "123321"
x = txt.isdecimal()
print(x) # expected output - True

# Work with Unicode
txt = "\u0030" # Unicode of 0
x = txt.isdecimal()
print(x) # exptected output - True
txt = "\u0047" # Unicode of 'G'
x = txt.isdecimal()
print(x) # exptected output - False becuase Unicode kisi deciaml number ka nhi Character ka hai
txt = "\u0033"
x = txt.isdecimal()
print(x)

# Negaitve par work karega yanahi
# Float par work karega nahi
txt1 = "-123"
txt2 = "1.3"
print(txt1.isdecimal())
print(txt2.isdecimal())

# Exponents par
txt = "\u00B2" # unicode for ²
x = txt.isdecimal()
print(x)
```

```python
# 20. isdigit() - REturn True if all are digit.
# Exponents like powers ^2/^3, etc. are also digits.
# Unke unicode par bhi work karega
# no parameter
txt = "1232"
x = txt.isdigit()
print(x)
txt = "as1232"
x = txt.isdigit()
print(x)

txt = "\u00B2" # unicode for ²
x = txt.isdigit()
print(x)

# Kya negative par work karega? 
txt = "-123"
x = txt.isdigit()
print(x)
# Decimal par
txt = "1.23"
x = txt.isdigit()
print(x)
```

```python
# 21. isidentifier() - Check karta hai ki Valid identifier hai ya nahi.
# True return karega ar hai toh.
# No parameter
# Identifiers wo hai to ji name dene liye use hote hai. lIke aap  python ke variables/functions/classes  ko name dete hai. Real world entities me aap ka name.
# But difference ye hota hai ki aap ke name me white spaces ho sakte hai. Lekin pyhton ke apnda nahi hote waha underscore (_) hote hai.
# Computer languager ke identifiers me only characters [a-z,A-Z], numbers [0-9] and underscore (_). Numbers se kabhi koi identifier start nahi hota.
txt = "Print"
x = txt.isidentifier()
print(x)
txt = "_MyName"
x = txt.isidentifier()
print(x)
x = "0Employee_Name".isidentifier()
print(x)
x = "Email_@0".isidentifier()
print(x)
```

```python
# 22. isnumeric() - Return True agar string me sare characters numbers hai toh.
# No paramter
txt = "123123"
x = txt.isnumeric()
print(x)

# Kya unicode keliye bhi work karta hai?
# Yes,check karte hai.
txt = "\u0030" # Unicode of 0
x = txt.isnumeric()
print(x)

# Kya exponents par bhi work karta hai?
# Yes, check karte hai. (² and ¾ dono par work karega)
txt = "\u00B2" # unicode for ²
x = txt.isnumeric()
print(x)

# Kya -ve and decimal par work karega?
# No, Check karte hai.
txt = "-1"
x = txt.isnumeric()
print(x) # because "-" is charcter
txt = "5.2"
x = txt.isnumeric()
print(x) # because "." is charcter
```

```python
# 33. maketrans() - Character/String ko change karne keliye use hota hai.
# *Important* - Ye 1 mapping table (basically dictionary object) return karta hai, Jo baad me translate() me dal kar aap value ko replace karte hai.
# 3 parameter leta hai.
# sysntax - string.maketrans(x,y,z)

txt = "Hi Arman-Man!"
x = "aAnmr" 
y = "eSrpu"

new_t = txt.maketrans(
    x, # REQUIRED. Agar Aap only x ko define karrahe ho toh wo necessary hai ki 1 dictionary object ho jo replace value table ho. 
    # Agar x ke alawa mulitple values hai toh x ki jagha wo value do jo replace karni hai.
    y # optional. second string jisse replace karna hai
    )
x = txt.translate(new_t)
print(new_t)
print(x)

# With Three character
txt = "Good Morning Arman!"
x = "aAnmr" 
y = "eSrpu"
z = "odngor" # Other characters jo replace nahi remove karne hai 

new_t = txt.maketrans(x,y,z)
print(new_t)
x = txt.translate(new_t)
print(x)
```

```python
# 34. partition() - Split the whole string in tuple of three parts.
# 1st part - part of string before the specified string
# 2nd part - the specified string
# 3rd part - part of string after the specified string
# 1 parameter

txt = "A quick brown fox jumps over a lazy dog."
x = txt.partition("brown")
print(x)

txt = "A quick brown fox jumps brown over a lazy dog."
x = txt.partition("brown")
print(x)
```

```python
# 35. rpartition() - Same as partition() but search for the last occurrence of the string.
# Split in 3 parts same as partition() but from the last.
x = txt.rpartition("brown")
print(x)
```

```python
# 36. split() - Split the whole string into list. 
# 2 parameters 
# 1st parameter - separator. Default " "
# 2nd parameter - maxsplit. Means count menas at how many occurrence you want to split  

# default parameter
txt = "My name is Arman Khan."
print(txt.split())

# custom parameter
txt = "armankhan52210@gmail.com"
print(txt.split("@"))
txt = "armankhan52210@gmail.com"
print(txt.split("@.")) # esi situation me nahi karega

# both custom parameter
txt = "My#name#is#Arman#Khan."
print(txt.split("#",2)) # 2 # tak hi split kara then stop kardiya
```

```python
# 37. rsplit() - Same as split() but start splitting from the right.
# 2 parameters

# default parameter
txt = "My name is Arman Khan."
print(txt.rsplit())

# custom parameter
txt = "armankhan52210@gmail.com"
print(txt.rsplit("@"))
txt = "armankhan52210@gmail.com"
print(txt.rsplit("@.")) # esi situation me nahi karega

# both custom parameter
txt = "My#name#is#Arman#Khan."
print(txt.rsplit("#",2)) # 2 # tak hi split kara then stop kardiya
```

```python
# 38. splitlines() - Split from the newline character (\n)
# 1 parameter (OPTIONAL)
# parameter naem - keeplinebreak - True/False if want to keep line break/ not
txt = "Hello, My name is Arman Khan.\nI am a Data Analyst"
x = txt.splitlines(
    False # Default
)
print(x)

txt = "Hello, My name is Arman Khan.\nI am a Data Analyst"
x = txt.splitlines(True)
print(x)

txt = "Hello.\nMy name is Arman Khan.\nI am a Data Analyst."
x = txt.splitlines(False)
print(x)
```

```python
# 42. translate() - Replace the spcified string/character.
# Takes an mapping table argument. maketrasn()  use karke mapping table de sakte ho.
# Agar maketrans() use nahi karte toh mapping table (dict) do lekin usem values ascii me honi chahiye na ki 'characters me'.

txt = "Hello, Arman!"
x = "Arman"
y = "Super"
table = txt.maketrans(x,y)
print(txt.translate(table))

table = {65: 83, 114: 117, 109: 112, 97: 101, 110: 114}
txt = "My name is Arman Khan."
print(txt.translate(table))
```

```python
# 44. format() - f"" string ki tarha work karta hai. String me specified value ko insert karta hai.
# Value that gonna be insert unko format bhi karta hai.
# 1 parameter leta hai.
# Lekin parameters me list of values.
# OR
# key-value pairs me value leta hai.
# Insert values with help of {} braces. You can also specify the index number.

# list values
txt = "Hello, My name is {}. My age is {}."
x = txt.format("Arman Khan","22")
print(x)

txt = "Hello, My name is {0}. My age is {1}."
x = txt.format("Arman Khan","22")
print(x)

txt = "Hello, My name is {1}. My age is {0}."
x = txt.format("Arman Khan","22")
print(x)

# key-value pair values
txt = "Hello, My name is {name}. My age is {age}."
x = txt.format(name= "Arman Khan", age = 22)
print(x)

txt = "Hello, My name is {age}. My age is {name}."
x = txt.format(name= "Arman Khan", age = 22)
print(x)
```

```python
print("Hello, My age is {:E} (scientific format with upper case e)".format(22)) 
print("Hello, My age is {:E} (scientific format with upper case e)".format(+22)) 
print("Hello, My age is {:E} (scientific format with upper case e)".format(-22)) 
```

```python
print("Hello, My age is {:.2f} (fix point number format)".format(22)) 
print("Hello, My age is {:f} (fix point number format)".format(+22)) 
print("Hello, My age is {:.3f} (fix point number format)".format(-22)) 
```

```python
print("Hello, My age is {:.02F} (scientific format with upper case e)".format(22)) 
print("Hello, My age is {:F} (scientific format with upper case e)".format(+22)) 
print("Hello, My age is {:.02F} (scientific format with upper case e)".format(-22)) 
print("Hello, My age is {:-.02F} (scientific format with upper case e)".format(-22)) # ye alag hai yaha hum 2 alag alag formatting apply kar rahe hai
```

```python
value1 = 123456.789
value2 = 0.000000123
value3 = float('inf')

# Using f-strings
print("g format (value1): {:g}".format(value1))
print("G format (value1): {:G}".format(value2))
print("g format (value2): {:g}".format(value1))
print("G format (value2): {:G}".format(value2))
print("G format (infinity): {:G}".format(value2))
print("g with precision (value1): {:.5g}".format(value1)) # 5 significant digit
print("g with alternate form (integer): {12:#g}") # adds decimal point

# Using str.format()
print("g format (value1): {0:g}".format(value1))
```

```python
txt = "The octal version of {0} is {0:o}"
print(txt.format(10))
```

```python
txt = "The Hexadecimal version of {0} is {0:x}"
print(txt.format(255))
```

```python
txt = "The Hexadecimal version of {0} is {0:X}"
print(txt.format(255))
```

```python
#Use "%" to convert the number into a percentage format:
txt = "You scored {:%}"
print(txt.format(0.25))

#Or, without any decimals:
txt = "You scored {:.0%}"
print(txt.format(0.25))
```

```python
# 45. format_map() - Same as format but you pass a real dictionary object.
# Good to use and preffered.

mapping = {
    "name" : "Arman",
    "age" : "22",
    "graduation" : "BCA",
    "address" : "somewhere in world".capitalize()
}
txt = "Hello, My name is {name}. I am {age} old. I am {graduation} Graduate. I lives in {address}."
print(txt.format_map(mapping))
```

```python
data = {"price": 49.99123}

text = "Price: ₹{price:.3f}"
print(text.format_map(data))
```
